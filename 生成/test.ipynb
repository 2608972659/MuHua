```python
import json
import os
system_prompt_filepath = os.path.join("prompts", "健康生活方式.json")
with open(system_prompt_filepath, "r", encoding="utf-8") as f:
    system_prompt = json.load(f)

prompt = []
patient_profile = "患者概况"
rag_context = "RAG检索结果"
role_prompt = "\n".join(system_prompt['role'])
prompt.append(role_prompt)
patient_profile_prompt = "\n".join(system_prompt['patient_profile']).format(patient_profile=patient_profile)
prompt.append(patient_profile_prompt)
input_suggestions_prompt = "\n".join(system_prompt['input_suggestions']).format(suggestions=rag_context)
prompt.append(input_suggestions_prompt)
output_format_prompt = "\n".join(system_prompt['output_format'])
prompt.append(output_format_prompt)
```

```python
print(prompt)
```
